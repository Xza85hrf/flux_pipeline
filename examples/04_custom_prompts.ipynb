## Prompt Anatomy

A well-structured prompt typically includes:
1. **Subject**: What you want (person, object, scene)
2. **Action**: What the subject is doing
3. **Setting**: Where the scene takes place
4. **Style**: Artistic style or mood
5. **Details**: Specific characteristics, lighting, atmosphere

```python
# Compare basic vs detailed prompts
prompts = {
    "Basic": "a cat",
    "Better": "a fluffy orange cat sitting on a windowsill",
    "Best": "a fluffy orange tabby cat with green eyes sitting on a sunlit windowsill, looking out at a garden, soft natural lighting, warm atmosphere, photorealistic"
}

results = {}
fixed_seed = 42

for level, prompt in prompts.items():
    print(f"{level}: {prompt}")
    image, seed = pipeline.generate_image(
        prompt=prompt,
        seed=fixed_seed,
        height=512,
        width=512,
        num_inference_steps=4
    )
    if image:
        results[level] = image
        print(f"  ✅ Generated\n")

# Display comparison
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
for idx, (level, image) in enumerate(results.items()):
    axes[idx].imshow(image)
    axes[idx].axis('off')
    axes[idx].set_title(level, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Style Modifiers

Add artistic style to your prompts.

```python
base_scene = "a castle on a hill at sunset"

styles = [
    "photorealistic, highly detailed",
    "oil painting, impressionist style",
    "watercolor painting, soft colors",
    "digital art, vibrant colors, fantasy art",
    "pencil sketch, black and white",
    "anime style, studio ghibli"
]

print(f"Base scene: {base_scene}\n")

style_results = []
for style in styles:
    prompt = f"{base_scene}, {style}"
    print(f"Style: {style}")
    
    image, seed = pipeline.generate_image(
        prompt=prompt,
        height=512,
        width=512,
        num_inference_steps=4,
        seed_profile=SeedProfile.BALANCED
    )
    
    if image:
        style_results.append((style.split(',')[0], image))
        print(f"  ✅ Done\n")

# Display style variations
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for idx, (style_name, image) in enumerate(style_results):
    axes[idx].imshow(image)
    axes[idx].axis('off')
    axes[idx].set_title(style_name, fontsize=12)

plt.suptitle(base_scene, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Lighting and Atmosphere

Lighting dramatically affects mood and quality.

```python
base_subject = "a mysterious forest path"

lighting_options = [
    "golden hour lighting, warm sunset glow",
    "soft morning mist, diffused light",
    "dramatic storm lighting, dark clouds",
    "moonlight filtering through trees, ethereal",
]

lighting_results = []
for lighting in lighting_options:
    prompt = f"{base_subject}, {lighting}"
    print(f"{prompt[:60]}...")
    
    image, _ = pipeline.generate_image(
        prompt=prompt,
        height=768,
        width=512,
        num_inference_steps=4
    )
    
    if image:
        lighting_results.append((lighting.split(',')[0], image))

# Display lighting comparison
fig, axes = plt.subplots(2, 2, figsize=(12, 16))
axes = axes.flatten()

for idx, (light, image) in enumerate(lighting_results):
    axes[idx].imshow(image)
    axes[idx].axis('off')
    axes[idx].set_title(light, fontsize=11)

plt.tight_layout()
plt.show()
```

## Quality Boosters

Keywords that improve output quality.

```python
quality_keywords = [
    "highly detailed",
    "8k resolution",
    "professional photography",
    "award winning",
    "masterpiece",
    "intricate details",
    "sharp focus",
    "cinematic composition"
]

print("Quality-boosting keywords:")
for kw in quality_keywords:
    print(f"  • {kw}")

# Example with quality keywords
basic_prompt = "a dragon"
enhanced_prompt = f"a majestic dragon, highly detailed, 8k resolution, professional digital art, award winning, intricate scales, sharp focus, cinematic composition"

print(f"\nBasic: {basic_prompt}")
print(f"Enhanced: {enhanced_prompt[:80]}...")
```

## Common Patterns

Proven prompt formulas.

```python
# Pattern 1: [Subject] + [Action] + [Setting] + [Style] + [Quality]
prompt1 = "A wise old wizard reading an ancient spellbook in a candle-lit library, fantasy art style, highly detailed, magical atmosphere"

# Pattern 2: [Scene] + [Mood] + [Lighting] + [Technical]
prompt2 = "An abandoned spaceship drifting in deep space, eerie and mysterious mood, dramatic rim lighting, cinematic composition, 8k detail"

# Pattern 3: [Character] + [Environment] + [Atmosphere] + [Style]
prompt3 = "A cyberpunk hacker in a neon-lit underground market, rainy night atmosphere, futuristic tech noir style, photorealistic"

pattern_prompts = [prompt1, prompt2, prompt3]
pattern_results = []

for idx, prompt in enumerate(pattern_prompts, 1):
    print(f"Pattern {idx}: {prompt[:60]}...")
    image, _ = pipeline.generate_image(
        prompt=prompt,
        height=768,
        width=768,
        num_inference_steps=4
    )
    if image:
        pattern_results.append(image)

# Display patterns
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
for idx, image in enumerate(pattern_results):
    axes[idx].imshow(image)
    axes[idx].axis('off')
    axes[idx].set_title(f"Pattern {idx+1}", fontsize=12)
plt.tight_layout()
plt.show()
```

## Best Practices

### DO:
- Be specific and descriptive
- Include lighting and atmosphere
- Mention artistic style or medium
- Use quality keywords
- Structure prompts logically

### DON'T:
- Be too vague ("a nice picture")
- Use contradictory terms
- Overcomplicate (keep it readable)
- Expect exact reproductions
- Forget about composition
